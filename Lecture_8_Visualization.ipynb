## Visualization

```python
import numpy as np
import pandas as pd 
```

```python
df_rfm = pd.read_csv('rfm.csv')
df_rfm.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sc = plt.scatter(
    df_rfm['R_score'],
    df_rfm['F_score'],
    c=df_rfm['M_score'],   # màu theo M_score
)

plt.xlabel('R_score')
plt.ylabel('F_score')
plt.title('RFM Scatter Plot (Color = M_score)')

# thêm colorbar
plt.colorbar(sc, label='M_score')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df_rfm,
    x='R_score',
    y='F_score',
    hue='M_score',      # màu theo M_score
    palette='viridis',
    s=100               # kích thước điểm
)

plt.title('RFM Scatter Plot')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sc = plt.scatter(
    df_rfm['Recency'],
    df_rfm['Frequency'],
    c=df_rfm['Monetary']   # màu theo Monetary
)

plt.xlabel('Recency')
plt.ylabel('Frequency')
plt.title('RFM Scatter Plot (Color = Monetary)')

plt.colorbar(sc, label='Monetary')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df_rfm,
    x='Recency',
    y='Frequency',
    hue='Monetary',
    palette='viridis',
    s=100
)

plt.title('RFM Scatter Plot')
plt.show()
```
