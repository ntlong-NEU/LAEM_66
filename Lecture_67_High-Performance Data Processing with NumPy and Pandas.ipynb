# PART 1: NUMPY - THE COMPUTATIONAL ENGINE
NumPy (Numerical Python) is the foundational package for scientific computing in Python. Almost every data science ecosystem (Pandas, Scikit-learn, TensorFlow) relies heavily on NumPy's `ndarray` (N-dimensional array) under the hood.

## 1.1 Environment Setup & Library Import
Let's import the libraries. We use standard aliases (`np` and `pd`) which are universally accepted in the data science community.

```python
import numpy as np
import pandas as pd
import time     # For benchmarking performance
import datetime as dt # For time-series operations
import warnings
warnings.filterwarnings('ignore') # Ignore non-critical warnings for cleaner output

print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {pd.__version__}")
print("Environment is ready!")
```

## 1.2 The Need for Speed: Python Lists vs. NumPy Arrays
Why learn NumPy when Python already has `lists`? 
- Python lists are dynamic arrays. They hold *pointers* to objects scattered across the computer's memory. This flexibility makes them slow.
- NumPy arrays (`ndarray`) contain data of a *single type* (homogeneous), stored in *contiguous* memory blocks. This allows the CPU to process them incredibly fast using C-level loops.

Let's prove this with a benchmark test: **Multiplying 5 million numbers.**

```python
SIZE = 5000000

# -----------------------------------------
# APPROACH 1: Standard Python List & For-Loop
# -----------------------------------------
py_list = list(range(SIZE))

start_time = time.time()
py_list = [x * 2 for x in py_list] # List comprehension
python_duration = time.time() - start_time
print(f"Python List Processing Time: {python_duration:.4f} seconds")

# -----------------------------------------
# APPROACH 2: NumPy Vectorization
# -----------------------------------------
np_array = np.arange(SIZE)

start_time = time.time()
np_array = np_array * 2 # Vectorized multiplication
numpy_duration = time.time() - start_time
print(f"NumPy Array Processing Time: {numpy_duration:.4f} seconds")

# -----------------------------------------
# CONCLUSION
# -----------------------------------------
speedup = python_duration / numpy_duration
print(f"\n🚀 NumPy is approximately {speedup:.2f} times faster than standard Python!")
```

## 1.3 Creating Arrays and Understanding Data Types (`dtype`)
Data types matter in Business Analytics. Storing data as `float64` takes twice as much RAM as `float32`. Managing memory efficiently is crucial when working with millions of rows.

```python
# Creating a 1D Array (Vector) representing Quantities Sold
quantities = np.array([10, 50, 120, 5, 200, 15, 80, 0, 12])
print("Quantities Array:", quantities)
print("Dimensions (ndim):", quantities.ndim)
print("Shape:", quantities.shape) # Notice the comma (9,) meaning 1D array of 9 elements
print("Data Type (dtype):", quantities.dtype)

# Creating an array with a specific data type to save memory
prices = np.array([2.5, 5.0, 10.0, 15.0, 2.0, 4.5, 8.0, 9.99, 1.5], dtype=np.float32)
print("\nPrices Array:", prices)
print("Prices Data Type:", prices.dtype)
```

## 1.4 Vectorization and Broadcasting in Business Context
**Vectorization** means applying a function to a whole array rather than looping through elements.
**Broadcasting** is a powerful mechanism that allows NumPy to work with arrays of different shapes when performing arithmetic operations (e.g., Array * Scalar).

```python
# Vectorization: Calculate Revenue (Quantity * Price)
revenues = quantities * prices
print("Base Revenues:\n", revenues)

# Broadcasting Example 1: Applying a flat shipping fee of $5 to all orders
revenues_with_shipping = revenues + 5.0 
print("\nRevenues + $5 Shipping:\n", revenues_with_shipping)

# Broadcasting Example 2: Applying a 10% VAT tax rate
final_revenues = revenues_with_shipping * 1.10
print("\nFinal Revenues (incl. VAT):\n", np.round(final_revenues, 2))
```

## 1.5 Statistical Aggregation
NumPy provides built-in methods to calculate descriptive statistics across arrays instantly.

```python
print("--- FINANCIAL SUMMARY ---")
print(f"Total Revenue (Sum): ${np.sum(final_revenues):.2f}")
print(f"Average Transaction (Mean): ${np.mean(final_revenues):.2f}")
print(f"Revenue Volatility (Std Dev): ${np.std(final_revenues):.2f}")

# Percentiles are crucial for finding outliers or medians
print(f"Median Transaction (50th Pctl): ${np.median(final_revenues):.2f}")
print(f"Top 10% Transactions start at: ${np.percentile(final_revenues, 90):.2f}")

# Min, Max and Argmax (Finding the Index)
print(f"\nHighest single revenue: ${np.max(final_revenues):.2f}")
best_idx = np.argmax(final_revenues)
print(f"The highest revenue occurred at index [{best_idx}], where quantity was {quantities[best_idx]} and price was {prices[best_idx]}")
```

## 1.6 Boolean Indexing (Masking) - Filtering without 'IF'
Instead of writing `if x > 100`, we create a "Mask" (an array of True/False) and pass it back into the array to filter data. This is foundational for Pandas.

```python
# Find "Wholesale" orders where Quantity > 100
wholesale_mask = quantities > 100
print("Boolean Mask Array:\n", wholesale_mask)

# Apply the mask to extract actual data
wholesale_quantities = quantities[wholesale_mask]
wholesale_revenues = final_revenues[wholesale_mask]
print("\nExtracted Wholesale Quantities:", wholesale_quantities)
print("Extracted Wholesale Revenues:", wholesale_revenues)

# Multiple conditions using Bitwise Operators (& for AND, | for OR)
# CRITICAL RULE: Always wrap conditions in parentheses ()
premium_deals_mask = (quantities >= 50) & (prices >= 5.0)
print("\nPremium Deals (Qty >= 50 AND Price >= $5):", final_revenues[premium_deals_mask])
```

## 2.2 Rigorous Data Cleaning Workflow
A Data Scientist spends 60-80% of their time cleaning data. Let's build a robust cleaning pipeline step-by-step.

```python
df_clean = df.copy() # Always work on a copy to avoid destroying raw data

# --------------------------------------------------------
# STEP 1: Handle Missing Values (NaN)
# --------------------------------------------------------
print("Missing values per column:\n", df_clean.isna().sum())

# In retail analytics, a transaction without a CustomerID cannot be tracked for RFM or behavior modeling.
# Decision: Drop rows where CustomerID is missing.
df_clean = df_clean.dropna(subset=['CustomerID'])
print(f"\nShape after dropping NaN CustomerIDs: {df_clean.shape}")
```

```python
# --------------------------------------------------------
# STEP 2: Handle Anomalies (Returns & Free Items)
# --------------------------------------------------------
# The dataset contains negative quantities. In this business context, InvoiceNo starting with 'C' means Cancelled.
# Let's identify them first.
returns = df_clean[df_clean['Quantity'] < 0]
print("\nSample of Returns/Cancellations:")
display(returns[['InvoiceNo', 'Quantity', 'UnitPrice']])

# To calculate pure Revenue, we filter out returns (Qty > 0) and zero-price items (UnitPrice > 0)
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['UnitPrice'] > 0)]
print(f"Shape after keeping only positive transactions: {df_clean.shape}")
```

```python
# --------------------------------------------------------
# STEP 3: Handle System Duplicates
# --------------------------------------------------------
# E-commerce systems frequently glitch and record double entries.
duplicate_mask = df_clean.duplicated(keep='first')
print(f"\nFound {duplicate_mask.sum()} duplicate rows.")

df_clean = df_clean.drop_duplicates()
print(f"Shape after dropping duplicates: {df_clean.shape}")
```

```python
# --------------------------------------------------------
# STEP 4: String Manipulation & Text Cleaning
# --------------------------------------------------------
# Product descriptions are messy, containing extra spaces and mixed cases.
print("\nBefore String Clean:\n", df_clean['Description'].tail(2).values)

# Use the .str accessor to apply methods to the entire column
df_clean['Description'] = df_clean['Description'].str.lower().str.strip()
print("After String Clean:\n", df_clean['Description'].tail(2).values)
```

```python
# --------------------------------------------------------
# STEP 5: Datetime Conversion
# --------------------------------------------------------
# InvoiceDate is currently an Object (String). We must cast it to a Datetime64 object for time-series analysis.
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

print("\n--- CLEANING COMPLETE. FINAL INFO ---")
df_clean.info()
display(df_clean)
```

⚠️ **Common Pitfall - SettingWithCopyWarning**

## MODULE 6: FEATURE ENGINEERING & ADVANCED SELECTION
Creating new dimensions out of existing data to empower marketing decisions.

```python
# 1. Calculate the Target Variable: Revenue
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']

# 2. Decompose Datetime using the .dt accessor
df_clean['Year'] = df_clean['InvoiceDate'].dt.year
df_clean['Month'] = df_clean['InvoiceDate'].dt.month
df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour
df_clean['DayOfWeek'] = df_clean['InvoiceDate'].dt.day_name()

# 3. Conditional Logic: np.where (Similar to IF in Excel)
df_clean['Is_High_Value'] = np.where(df_clean['Revenue'] > 50, 'Yes', 'No')

# Display engineered features
display(df_clean[['InvoiceNo', 'Revenue', 'DayOfWeek', 'Hour', 'Is_High_Value']].head())
```

### 2.3 Selecting Data with `.loc` and `.iloc`
- `.iloc[rows, cols]`: Selects by **Integer Position** (e.g., Row 0, Column 1)
- `.loc[row_labels, col_labels]`: Selects by **Labels or Conditions**.

```python
# Example using .loc to filter High Value orders in the UK, selecting specific columns
high_val_uk = df_clean.loc[
    (df_clean['Country'] == 'United Kingdom') & (df_clean['Is_High_Value'] == 'Yes'), 
    ['CustomerID', 'Description', 'Revenue']
]
print("High Value UK Orders:")
display(high_val_uk)
```

## MODULE 7: AGGREGATION (SPLIT-APPLY-COMBINE)
The `groupby()` function is the workhorse of Pandas. It groups data by a category, applies a function (sum, mean, max), and combines the results.

```python
# 1. Single Aggregation: Total Revenue by Country
country_rev = df_clean.groupby('Country')['Revenue'].sum().sort_values(ascending=False)
print("--- Revenue by Country ---")
print(country_rev)

# 2. Multiple Aggregations using .agg()
# We want to know: For each country, what is the Total Revenue, Average Order Value, and Unique Customer Count?
country_kpis = df_clean.groupby('Country').agg({
    'Revenue': ['sum', 'mean'],     # Apply multiple functions to one column
    'CustomerID': 'nunique'         # Count unique customers
})
print("\n--- Advanced Country KPIs ---")
display(country_kpis)
```

### 2.4 Pivot Tables
Just like Excel, Pandas can create multidimensional matrices.

```python
pivot = pd.pivot_table(
    df_clean, 
    values='Revenue', 
    index='Country',        # Rows
    columns='Is_High_Value',# Columns
    aggfunc='sum',          # Aggregation method
    fill_value=0            # Replace NaN with 0
)
print("--- Pivot Table: Revenue Matrix ---")
display(pivot)
```

## MODULE 8: CAPSTONE 1 - RFM MARKETING ANALYSIS
RFM modeling evaluates customer value based on three metrics:
- **Recency (R):** Days since last purchase (Lower is better).
- **Frequency (F):** Total number of transactions (Higher is better).
- **Monetary (M):** Total money spent (Higher is better).

```python
# Step 1: Define a snapshot date. Usually the day after the last transaction in your dataset.
snapshot_date = df_clean['InvoiceDate'].max() + dt.timedelta(days=1)

# Step 2: Group by CustomerID and calculate RFM metrics
rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days, # Recency
    'InvoiceNo': 'nunique',                                  # Frequency
    'Revenue': 'sum'                                         # Monetary
})

# Step 3: Rename columns
rfm.columns = ['Recency', 'Frequency', 'Monetary']

print("--- Raw RFM Metrics ---")
display(rfm)
```

## MODULE 9: CAPSTONE 2 - RELATIONAL DATA WRANGLING (UCI)
In the real world, data isn't in one flat file. It's stored in normalized SQL databases. We must use `pd.merge()` to recreate the unified view.

```python
# Let's generate a comprehensive set of relational tables representing a Restaurant App.

# Table 1: User Profiles (Demographics)
users = pd.DataFrame({
    'userID': ['U101', 'U102', 'U103', 'U104', 'U105'],
    'budget': ['high', 'low', 'medium', 'high', 'medium'],
    'smoker': ['yes', 'no', 'no', 'yes', 'no']
})

# Table 2: Restaurants (Geoplaces)
places = pd.DataFrame({
    'placeID': ['P01', 'P02', 'P03', 'P04', 'P99'], # Notice P99 has no ratings later
    'name': ['El Gaucho', 'KFC', 'Sushi Hokkaido', 'Pizza 4P', 'Ghost Kitchen'],
    'city': ['Hanoi', 'HCMC', 'Da Nang', 'Hanoi', 'Hanoi']
})

# Table 3: Ratings (The Transactional Table linking Users and Places)
ratings = pd.DataFrame({
    'userID': ['U101', 'U101', 'U102', 'U103', 'U104', 'U104'],
    'placeID': ['P01', 'P03', 'P02', 'P03', 'P01', 'P04'],
    'food_rating': [5, 4, 3, 5, 4, 5],
    'service_rating': [5, 3, 2, 4, 5, 4]
})

print("Database initialized.")
```

### 3.1 Understanding Joins
- **Inner Join:** Keeps only rows where the keys exist in BOTH tables.
- **Left Join:** Keeps ALL rows from the Left table, and matches from the Right. Fills NaN if no match is found.

```python
# --------------------------------------------------
# INNER JOIN WORKFLOW: Building a Recommendation Matrix
# --------------------------------------------------
# Step 1: Merge Ratings with Users
merged_step1 = pd.merge(ratings, users, on='userID', how='inner')

# Step 2: Merge the result with Places
full_matrix = pd.merge(merged_step1, places, on='placeID', how='inner')

# Feature Engineering
full_matrix['Total_Score'] = full_matrix['food_rating'] + full_matrix['service_rating']

print("--- Full Inner Joined Matrix ---")
display(full_matrix)
```

```python
# --------------------------------------------------
# LEFT JOIN WORKFLOW: Finding Orphan Data
# --------------------------------------------------
# Business Question: Are there any restaurants in our database that have ZERO ratings?
# We use a Left Join. Places is the Left table. Ratings is the Right table.

places_with_ratings = pd.merge(places, ratings, on='placeID', how='left')

print("--- Left Join Results ---")
display(places_with_ratings)

# Find the orphan restaurant by looking for NaNs in the rating columns
orphan_restaurants = places_with_ratings[places_with_ratings['food_rating'].isna()]
print("\n--- Restaurants with Zero Ratings ---")
display(orphan_restaurants[['name', 'city']])
```

### 3.2 Extracting Business Insights from Merged Data
Now that we have `full_matrix`, we can answer complex business questions.

```python
# Question: What is the highest rated restaurant among "High Budget" users?
high_budget_ratings = full_matrix[full_matrix['budget'] == 'high']

top_places = high_budget_ratings.groupby('name')['Total_Score'].mean().sort_values(ascending=False)
print("Top Restaurants for High Budget Users:")
print(top_places)
```
