```python
# count frequency:
#1 normalization
foh = foh.lower()
special_char = ['.',',','(',')','*','"','-','/',':','+','—']
for char in special_char:
    foh = foh.replace(char,' ')
print(foh) 
```

```python
#2 split:
words = foh.split()
print(words)
print(len(words))
```

```python
[1,2,1,1,1,2,3].count(1)
```

```python
#3 count: fre_words = [[w,fre_w]]
unique_words = set(words)
fre_words = []
for w in unique_words:
    fre_words.append([w,words.count(w)])
print(fre_words)
```

```python
sorted(fre_words, reverse= True, key= lambda x: x[1])
```

```python
words = ['aa','ab','abc','aa','aa','abc','abc','abc']
fre_dict = {}
for w in words:
    if w in fre_dict:
        fre_dict[w] += 1
    else:
        fre_dict[w] = 1
print(fre_dict)
```
