```python
def normlize_text(text:str)->str:
    '''get a norm_text'''
    special_char ={'.','/','-','_','*','!','='}
    text_nor = text.lower()
    for char in special_char:
        text_nor = text_nor.replace(char,' ')
    return text_nor
```
