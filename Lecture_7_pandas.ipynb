## Customer segmentation by RFM model

### RFM model
+ R: Recency. Tính bằng cách lấy ngày phân tích trừ đi ngày mua hàng
+ F: Frequency. Tính bằng số đơn (InvoiceNo)
+ M: Monetary. Tính bằng số tiền khách hàng đã chi tiêu. Tính tổng giá trị các đơn hàng mà khách hàng đã mua

```python
df_cs.head(10)
```

```python
# tính M,
df_M = df_cs[['CustomerID','TotalPrice']].groupby("CustomerID").sum().rename(columns={'TotalPrice': 'Monetary'})
```

```python
# tính F
df_F = df_cs[['CustomerID', 'InvoiceNo']].groupby('CustomerID').count().rename(columns={"InvoiceNo":"Frequency"})
```

```python
snapshot_date = df_cs['InvoiceDate'].max()+ pd.Timedelta(days=1)
```

```python
df_cs['days'] = (snapshot_date - df_cs['InvoiceDate']).dt.days
df_cs.head()
```

```python
# tính R
df_R = df_cs[['CustomerID','days']].groupby('CustomerID').min().rename(columns={'days':'Recency'})
```

```python
df_R.head()
```

```python
df_F.head()
```

```python
pd.merge(df_R,df_F, left_index=True, right_index=True)
```

```python
rfm = df_R.join([df_F,df_M])
rfm.head()
```

```python

# Calculate R_score: lower recency is better, so reverse labels
rfm['R_score'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1])

# Calculate F_score: higher frequency is better
rfm['F_score'] = pd.qcut(rfm['Frequency'], 5, labels=[1, 2, 3, 4, 5])

# Calculate M_score: higher monetary is better
rfm['M_score'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5])

rfm.head()
```

```python
rfm.info()
```

```python
# Calculate RFM_score as the sum of R_score, F_score, and M_score
rfm['RFM_score'] = rfm['R_score'].astype(int) + rfm['F_score'].astype(int) + rfm['M_score'].astype(int)

# Define segments based on RFM_score
def segment_customer(score):
    if score >= 12:
        return 'High Value'
    elif score >= 9:
        return 'Medium Value'
    elif score >= 6:
        return 'Low Value'
    else:
        return 'Lost'

rfm['Segment'] = rfm['RFM_score'].apply(segment_customer)

rfm.head()
```

```python
rfm.to_csv('rfm.csv')
```
